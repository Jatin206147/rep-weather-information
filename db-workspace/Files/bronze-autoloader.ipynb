```python
df = spark.readStream.format("cloudFiles") \
    .option("cloudFiles.format","json") \
    .option("cloudFiles.schemaLocation", "/Volumes/ws_weather_informaton/bronze/weather_data_autoloader/schema_location/") \
    .option("pathGlobfilter", "*.json") \
    .option("cloudFiles.inferColumnTypes", "true") \
    .option("cloudFiles.schemaEvolutionMode","rescue") \
    .load("abfss://weather-data@stgweatherapp.dfs.core.windows.net/")
```

```python
df.writeStream.format("delta") \
    .option("mergeSchema", "true") \
    .option("checkpointLocation", "/Volumes/ws_weather_informaton/bronze/weather_data_autoloader/checkpoint_location/") \
    .trigger(availableNow=True) \
    .toTable("ws_weather_informaton.bronze.weather_data")
```
